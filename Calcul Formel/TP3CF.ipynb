## Suites

### Rappels de cours : 
- **Une suite est *récurrente* (ou *définie par récurrence*) si son terme général est exprimé en fonction des termes précédents.**  


- **La suite $(u_{n})$ est une *suite arithmétique* s'il existe un réel $r$ tel que : 
$ \forall n \in \mathbb{N}, u_{n+1} = u_{n} + r.$  
Le nombre $r$ est alors appelé la raison de la suite $(u_{n})$.**  


- **La suite $(u_{n})$ est une *suite géométrique* s'il existe un réel $q$ tel que : 
$ \forall n \in \mathbb{N}, u_{n+1} = q * u_{n} .$  
Le nombre $q$ est alors appelé la raison de la suite $(u_{n})$.**

### Exercice 3 (Suite de Fibonacci)
La suite de Fibonacci est définie par : 
$$ u_{0} = u_{1} = 1, \quad \mbox{et} \quad u_{n} = u_{n-1} + u_{n-2} \mbox{ pour } n\geq 2.$$
Cette suite représente par exemple l'évolution du nombre de couples de lapins chaque mois, si on place un couple de lapins dans un milieu clos. 
- Ecrire une fonction récursive qui calcule $u_n$ pour $n$ donné et calculer le 10ième terme de cette suite. 
- Modifier la fonction pour ne pas faire d'appels récursifs, en utilisant une approche itérative. 
- Comparer le temps mis par chacune des fonctions pour calculer le 30ième terme de cette suite. Pour cela, on utilisera la commande `timeit('...commande...')`.

```python
#récursive correction
reset()
def fibor(n):
    if n==0 or n==1:
        return 1
    else:
        return fibor(n-1)+fibor(n-2)
print(fibor(9))
```

```python
timeit('fibor(29)')
```

```python
#récursive
reset()
import timeit
n=var('n')
u=0

def fib(n):
    if n <= 1: return n
    return fib(n-1) + fib(n-2)

for n in range(2 ,31):
    u=u+n
    res= (u*(n-1)) + (u*(n-2))
    print("u",n,"=",res)
    temps_rec = timeit.timeit('fib(30)', globals=globals(), number=1)

print(f"Version Récursive : {temps_rec:} secondes")
```
