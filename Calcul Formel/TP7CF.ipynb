Un message étant donné en langue naturelle, le chiffrement ou cryptage consiste à le traduire en langage codé et le déchiffrement consiste à traduire le message codé en langage naturel.  
On choisit le chiffrement suivant :
à partir du nombre $x$, on calcule $11x+8$ puis le reste de la division euclidienne de $11x+8$ par 26 noté $y$ ($y$ est compris entre 0 et 25).
Un tel chiffrement est dit ''affine''.  
_Exemple :  
la lettre m est assimilée à $x=12$ : 11 * 12+8=140 et $140\equiv 10 \mod 26$. Ainsi m est codée k._
- Tester ce chiffrement. 

```python
L0 = list('message')
print(L0)
L = [ord(j)-97 for j in L0]
print(L)
#on ajoute ici le chiffrement y = 11x+8 mod 26
N=[(11*x+8)%26 for x in L ]
print(N)
M = [chr(n+97) for n in N]
print(M)
print(''.join(M))    
```

```python
def chiffre(m):
    L0 = list(m)
    L =[ord(j)-97 for j in L0]
    N = [(11*x+8)%26 for x in L ]
    M = [chr(n+97)for n in N ]
    return ''.join(M)
```

```python
chiffre('bonjour')
```

```python
def dechiffre(m):
    L0 = list(m)
    L =[ord(j)-97 for j in L0]
    N = [(19*(y-8))%26 for y in L]
    M = [chr(n+97)for n in N ]
    return ''.join(M)
```
